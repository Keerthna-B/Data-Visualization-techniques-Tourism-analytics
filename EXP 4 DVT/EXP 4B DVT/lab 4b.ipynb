```python
import pandas as pd
import matplotlib.pyplot as plt

# Import Dataset
df=pd.read_csv(r"C:\Users\Keertu\Downloads\tourism_dataset_5000.csv") 


# Dataset Inspection
print("===== FIRST FIVE RECORDS =====")
display(df.head())

print("\n===== SHAPE OF DATASET =====")
print(df.shape)

print("\n===== DATA TYPES =====")
print(df.dtypes)

print("\n===== SUMMARY =====")
display(df.describe())


# Filtering Data

# Tourists with Tourist Rating greater than 4
print("\n===== TOURISTS WITH RATING > 4 =====")
high_rating = df[df["Tourist Rating"] > 4]
display(high_rating.head())

# Tourists with high satisfaction
print("\n===== TOURISTS WITH SATISFACTION > 80 =====")
high_satisfaction = df[df["Satisfaction"] > 80]
display(high_satisfaction.head())


# Sorting Data
print("\n===== SORT BY TOURIST RATING (DESCENDING) =====")
sorted_df = df.sort_values(by="Tourist Rating", ascending=False)
display(sorted_df.head())


# Statistical Analysis
print("\n===== TOURIST RATING STATISTICS =====")
print("Mean :", df["Tourist Rating"].mean())
print("Median :", df["Tourist Rating"].median())
print("Mode :")
print(df["Tourist Rating"].mode())
print("Standard Deviation :", df["Tourist Rating"].std())

print("\n===== DESCRIPTIVE STATISTICS =====")
display(df.describe())


# Grouping and Aggregation
print("\n===== AVERAGE TOURIST RATING BY SITE =====")
site_avg = df.groupby("Site Name")["Tourist Rating"].mean()
display(site_avg)


# Correlation Analysis
print("\n===== CORRELATION =====")

corr = df.select_dtypes(include='number').corr()

plt.figure(figsize=(10, 8))

plt.imshow(corr)

plt.colorbar()

plt.xticks(
    range(len(corr.columns)),
    corr.columns,
    rotation=90
)

plt.yticks(
    range(len(corr.columns)),
    corr.columns
)

plt.title("Correlation Matrix")

plt.show()


# Visualization 1
site_avg.plot(kind="bar")

plt.title("Average Tourist Rating by Site")
plt.xlabel("Site Name")
plt.ylabel("Average Tourist Rating")
plt.show()


# Visualization 2
plt.figure(figsize=(8,5))

plt.scatter(df["Tour Duration"], df["Tourist Rating"])

plt.title("Tour Duration vs Tourist Rating")
plt.xlabel("Tour Duration")
plt.ylabel("Tourist Rating")

plt.show()
```
