```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv(r"C:\Users\Keertu\Downloads\tourism_dataset_5000.csv")

print("Shape:", df.shape)
```

```python


site_counts = df["Site Name"].value_counts()

plt.figure(figsize=(10,5))
plt.plot(site_counts.index, site_counts.values, marker='o')
plt.title("Number of Tourists by Site")
plt.xlabel("Site Name")
plt.ylabel("Number of Tourists")
plt.xticks(rotation=45)
plt.grid(True)
plt.show()
```

```python
site_counts = df["Site Name"].value_counts()

plt.figure(figsize=(7,5))
plt.bar(site_counts.index, site_counts.values)
plt.title("Tourists by Site")
plt.xlabel("Site Name")
plt.ylabel("Number of Tourists")
plt.xticks(rotation=45)
plt.show()
```

```python
import ast
df["Interests"] = df["Interests"].apply(ast.literal_eval)
all_interests = df["Interests"].explode()
interest_counts = all_interests.value_counts()
plt.figure(figsize=(8, 8))
plt.pie(
    interest_counts.values,
    labels=interest_counts.index,
    autopct='%1.1f%%',
    startangle=90
)
plt.title("Tourist Interest Distribution")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df["Age"], bins=30)
plt.title("Distribution of Tourist Age")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(
    df["Age"],
    df["Tourist Rating"],
    alpha=0.3
)

plt.title("Age vs Tourist Rating")
plt.xlabel("Age")
plt.ylabel("Tourist Rating")

plt.show()
     
```
